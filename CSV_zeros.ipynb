To manage zeros in csv files

```python
import pandas as pd
import os
import re
```

```python
# Set the directory containing the CSV files
input_doss = 'Fifa_Dataset'
```

```python
def sort_key(filename):
    match = re.search(r'keypoints_frame_(\d+)\.csv', filename)
    if match:
        return int(match.group(1))
    return filename

def f(directory):
    
    # Lire tous les fichiers CSV dans une liste de DataFrames
    fichiers_csv = sorted([f for f in os.listdir(directory) if f.endswith('.csv')], key=sort_key)
    dataframes = [pd.read_csv(os.path.join(directory, f)) for f in fichiers_csv]


    # Parcourir chaque fichier à partir du deuxième jusqu'à l'avant-dernier
    for idx in range(0, len(dataframes)):
    #for idx in range(13, 14):
        df_courant = dataframes[idx]
            
        for col in ['x','y']:
            for i in range(len(df_courant)):
                #print(int(df_courant.at[i, col]))
                if df_courant.at[i, col] == 0:
                    df_courant.at[i,'z'] = 0

        # Sauvegarder le fichier CSV modifié
        df_courant.to_csv(os.path.join(directory, fichiers_csv[idx]), index=False)


def verify(directory):
    
    # Lire tous les fichiers CSV dans une liste de DataFrames
    fichiers_csv = sorted([f for f in os.listdir(directory) if f.endswith('.csv')], key=sort_key)
    dataframes = [pd.read_csv(os.path.join(directory, f)) for f in fichiers_csv]

    for idx in range(0, len(dataframes)):
        for i in range(len(dataframes[idx])):
            if (dataframes[idx].at[i, 'x'] == 0 or dataframes[idx].at[i, 'y'] == 0) and dataframes[idx].at[i, 'z'] != 0 :
                print("error at ", directory, " at frame :", idx, ", at line :", i)


```

```python
for dire in os.listdir(input_doss) :
    if dire.startswith('output_'):
        directory = os.path.join(input_doss, dire)
        for pen in os.listdir(directory) :
            if pen.startswith('penalty_'):
                chemin_complet = os.path.join(directory, pen)
                f(chemin_complet)
                verify(chemin_complet)
```

# Processing and Verifying CSV Files

## Introduction

This script processes CSV files containing body joint coordinates, modifies specific values based on conditions, and verifies the modifications. It ensures that if a point's `x` or `y` coordinate is zero, its `z` value is set to zero. Additionally, it verifies the integrity of the data after modifications.

### Importing Necessary Libraries

```python

````python
import pandas as pd
import os
import re
````

### Setting the Directory Containing the CSV Files


````python
# Set the directory containing the CSV files
input_doss = 'Fifa_Dataset'
````

### Function to Sort Files by Frame Number


````python
def sort_key(filename):
    match = re.search(r'keypoints_frame_(\d+)\.csv', filename)
    if match:
        return int(match.group(1))
    return filename
````

### Function to Process CSV Files
This function reads all CSV files in the specified directory, checks each x and y coordinate, and sets the corresponding z value to zero if either x or y is zero. It then saves the modified CSV files.

````python
def f(directory):
    # Read all CSV files into a list of DataFrames
    csv_files = sorted([f for f in os.listdir(directory) if f.endswith('.csv')], key=sort_key)
    dataframes = [pd.read_csv(os.path.join(directory, f)) for f in csv_files]

    # Iterate over each file from the first to the last
    for idx in range(0, len(dataframes)):
        current_df = dataframes[idx]
        
        for col in ['x','y']:
            for i in range(len(current_df)):
                if current_df.at[i, col] == 0:
                    current_df.at[i, 'z'] = 0

        # Save the modified CSV file
        current_df.to_csv(os.path.join(directory, csv_files[idx]), index=False)
````

### Function to Verify Modifications
This function verifies that the modifications were correctly applied. If any x or y coordinate is zero and the corresponding z value is not zero, it prints an error message.

````python
def verify(directory):
    # Read all CSV files into a list of DataFrames
    csv_files = sorted([f for f in os.listdir(directory) if f.endswith('.csv')], key=sort_key)
    dataframes = [pd.read_csv(os.path.join(directory, f)) for f in csv_files]

    for idx in range(0, len(dataframes)):
        for i in range(len(dataframes[idx])):
            if (dataframes[idx].at[i, 'x'] == 0 or dataframes[idx].at[i, 'y'] == 0) and dataframes[idx].at[i, 'z'] != 0:
                print("Error at ", directory, " at frame:", idx, ", at line:", i)
````

### Main Processing Loop
The main loop iterates over the directories, processes each CSV file, and verifies the modifications.

````python
for dire in os.listdir(input_doss) :
    if dire.startswith('output_'):
        directory = os.path.join(input_doss, dire)
        for pen in os.listdir(directory) :
            if pen.startswith('penalty_'):
                chemin_complet = os.path.join(directory, pen)
                f(chemin_complet)
                verify(chemin_complet)
````

## Conclusion
This script processes and verifies CSV files containing body joint coordinates. It ensures data integrity by setting the z value to zero for points where either the x or y coordinate is zero and verifies the changes.
